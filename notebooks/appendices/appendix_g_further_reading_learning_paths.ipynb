# Appendix G: Further Reading and Learning Paths

Use the machine-readable routes to select, inspect, and track a learning plan.

**Level:** All levels  |  **Time:** 30 minutes.


```python
from pathlib import Path
import sys
root=Path.cwd().parents[1];sys.path.insert(0,str(root/'src'))
from datasciencebook.learning_paths import load_paths,available_paths,path_plan,plan_summary,check_progress
rows=load_paths(root/'project/learning_paths.csv')
print('Paths:',len(available_paths(rows)))
```

```python
for path_id,name in available_paths(rows).items():
    summary=plan_summary(rows,path_id)
    print(path_id,name,int(summary['hours']))
```

```python
plan=path_plan(rows,'analyst')
for row in plan: print(row['sequence'],row['resource'],'->',row['deliverable'])
```

```python
progress=check_progress(plan,['Chapters 1-8','Chapters 9-19'])
print(progress)
assert progress['next_resource']=='Chapters 20-26'
```

```python
import matplotlib.pyplot as plt
summaries=[plan_summary(rows,p) for p in available_paths(rows)]
names=[available_paths(rows)[s['path_id']] for s in summaries];hours=[s['hours'] for s in summaries]
fig,axes=plt.subplots(1,2,figsize=(12,4));axes[0].barh(names,hours,color='#38598b');axes[0].set(title='Estimated hours by path',xlabel='Hours')
stages=[r['stage'] for r in plan];stage_hours=[r['hours'] for r in plan];axes[1].bar(stages,stage_hours,color='#e09f3e');axes[1].set(title='Analyst path by stage',ylabel='Hours');axes[1].tick_params(axis='x',rotation=25);fig.tight_layout();plt.show()
```

## Interpretation

Hours help scope the route but do not certify competence. Advance when the deliverable passes its gate.
