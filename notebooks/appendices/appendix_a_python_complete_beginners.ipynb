```python
product='SAMBAL_250'
weekly_cases=84
price_per_case=6.5
revenue=weekly_cases*price_per_case
print(describe_value('product',product))
print(f'Revenue: {revenue:.2f}')
```

```python
records=[{'product':'tea','cases':8},{'product':'coffee','cases':12},{'product':'sambal','cases':16}]
high=filter_records(records,'cases',10)
print('Selected:',[row['product'] for row in high])
print('Mean cases:',safe_mean(row['cases'] for row in records))
```

```python
position=70
target=96
if position<target:
    order=target-position
else:
    order=0
print(f'Order: {order} cases')
print(case_summary(records))
```
