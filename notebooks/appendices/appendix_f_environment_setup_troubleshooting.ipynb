Inspect the active runtime without changing it.

**Level:** Beginner  |  **Time:** 30 minutes.


```python
from pathlib import Path
import sys,json
root=Path.cwd().parents[1];sys.path.insert(0,str(root/'src'))
from datasciencebook.environment import environment_report,readiness
print('Python:',sys.version_info.major,sys.version_info.minor)
```

```python
report=environment_report(('numpy','pandas','matplotlib','pytest'))
summary={'python':report['python'],'virtual_environment':bool(report['virtual_environment']),'git':bool(report['commands']['git']),'packages':report['packages']}
print(json.dumps(summary,indent=2))
```

```python
synthetic_ready={'virtual_environment':'/project/.venv','commands':{'git':'/usr/bin/git'},'packages':{'numpy':'2.0','pytest':'8.0'}}
synthetic_missing={'virtual_environment':None,'commands':{'git':'/usr/bin/git'},'packages':{'numpy':'2.0','pytest':None}}
print(readiness(synthetic_ready));print(readiness(synthetic_missing))
```

```python
checks=['repository root','expected branch','active environment','matching kernel','declared packages','data version','tests']
for number,item in enumerate(checks,1):print(f'{number}. {item}')
```

```python
import matplotlib.pyplot as plt
status=[True,True,False,True,True,True,False];colors=['#4f8a63' if x else '#b84b4b' for x in status]
fig,ax=plt.subplots(figsize=(9,4));ax.barh(checks,status,color=colors);ax.set(xlim=(0,1),xticks=[0,1],xticklabels=['Needs action','Ready'],title='Example environment readiness');fig.tight_layout();plt.show()
```

## Interpretation

The real report is observational and may show limitations. The synthetic examples demonstrate how readiness rules communicate specific actions without modifying the computer.
