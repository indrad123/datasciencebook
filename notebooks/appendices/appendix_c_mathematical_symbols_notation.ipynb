```python
from pathlib import Path
import sys, numpy as np, matplotlib.pyplot as plt
root=Path.cwd().parents[1];sys.path.insert(0,str(root/'src'))
from datasciencebook.notation import sigma,weighted_sum,dot,euclidean_norm,empirical_probability,central_difference
print('Notation helpers ready.')
```

```python
x=[82,90,87,95]
w=[0.1,0.2,0.3,0.4]
print('Sigma:',sigma(x))
print('Mean:',sigma(x)/len(x))
print('Weighted sum:',weighted_sum(x,w))
```

```python
features=[0.8,-0.5,1.2];weights=[1.5,0.4,-0.2]
print('Dot product:',round(dot(features,weights),2))
print('Feature norm:',round(euclidean_norm(features),3))
on_time=[True,True,False,True,True]
print('Empirical probability:',empirical_probability(on_time))
```

```python
loss=lambda theta:(theta-3)**2+2
print('Approximate derivative at 1:',round(central_difference(loss,1),3))
print('Approximate derivative at 3:',round(central_difference(loss,3),3))
```

```python
theta=np.linspace(0,6,121);values=(theta-3)**2+2
weekly=np.array(x,dtype=float)
fig,axes=plt.subplots(1,2,figsize=(10,4))
axes[0].plot(theta,values);axes[0].scatter([3],[2],color='#b84b4b');axes[0].set(title=r'$L(\theta)=(\theta-3)^2+2$',xlabel=r'$\theta$',ylabel='Loss')
axes[1].plot(np.arange(1,5),np.cumsum(weekly),marker='o',color='#4f8a63');axes[1].set(title=r'Cumulative $\sum_{i=1}^{t}x_i$',xlabel=r'Week $t$',ylabel='Cases')
fig.tight_layout();plt.show()
```
