# Appendix H: Statistical Reporting Checklist

Audit an illustrative NRG report before release.


```python
from pathlib import Path
import sys
root=Path.cwd().parents[1];sys.path.insert(0,str(root/'src'))
from datasciencebook.reporting import load_checklist,item_ids,evaluate_report
c=load_checklist(root/'project/statistical_reporting_checklist.json')
print(len(item_ids(c)))
```

```python
responses={i:True for i in item_ids(c)}
for missing in ['multiplicity_addressed','sensitivity_analysis_reported','owner_and_follow_up_named']:responses[missing]=False
r=evaluate_report(c,responses);print(r['complete'],r['total'],round(r['score'],2),r['ready'])
```

```python
for s in r['sections']:print(s['id'],f"{s['complete']}/{s['total']}")
```

```python
print('Next actions:');[print('-',x) for x in r['missing']]
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(8,4));ax.bar([s['id'] for s in r['sections']],[s['score'] for s in r['sections']],color='#3f6f8f');ax.set(ylim=(0,1),ylabel='Completion',title='Illustrative report checklist');plt.xticks(rotation=20);fig.tight_layout();plt.show()
```

A completion score detects omissions. It does not judge whether the report is scientifically correct.
