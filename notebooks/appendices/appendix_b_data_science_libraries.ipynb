```python
from pathlib import Path
import sys,numpy as np,pandas as pd,matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LinearRegression
root=Path.cwd().parents[1];sys.path.insert(0,str(root/'src'))
from datasciencebook.library_tools import array_report,frame_contract
print('Imports ready.')
```

```python
demand=np.array([[82,90,87,95],[61,65,68,72],[48,51,54,58]],dtype=float)
print(array_report(demand))
print('Weekly totals:',demand.sum(axis=0).astype(int).tolist())
```

```python
frame=pd.DataFrame({'week':[1,2,3,4],'sambal':demand[0],'promotion':[0,1,0,1]})
frame=frame.assign(cases_per_day=frame['sambal']/7)
print(frame_contract(frame,['week','sambal','promotion'],['week']))
print('Promotional mean:',round(frame.loc[frame['promotion']==1,'sambal'].mean(),1))
```

```python
X=frame[['week','promotion']];y=frame['sambal']
model=LinearRegression().fit(X,y);prediction=model.predict(pd.DataFrame({'week':[5],'promotion':[0]}))[0]
r,p=stats.pearsonr(frame['week'],frame['sambal'])
print(f'Week-5 prediction: {prediction:.1f}')
print(f'Correlation: {r:.3f}; p-value: {p:.3f}')
```

```python
fig,axes=plt.subplots(1,2,figsize=(10,4));axes[0].plot(frame['week'],frame['sambal'],marker='o');axes[0].set(title='Sambal demand',xlabel='Week',ylabel='Cases');axes[1].bar(['Sambal','Coffee','Tea'],demand.mean(axis=1),color=['#b84b4b','#795548','#4f8a63']);axes[1].set(title='Mean weekly demand',ylabel='Cases');fig.tight_layout();plt.show()
```

## Interpretation

NumPy represents the product-by-week matrix, pandas adds labels, the helper validates the table contract, scikit-learn fits a predictive model, SciPy calculates a statistical association, and Matplotlib communicates results. Four observations are far too few for a production conclusion.
