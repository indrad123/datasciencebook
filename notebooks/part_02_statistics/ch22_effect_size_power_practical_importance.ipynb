# Chapter 22: Effect Size, Power, and Practical Importance
Synthetic NRG examples connect statistical magnitude to operational decisions.

```python
import matplotlib.pyplot as plt
from datasciencebook.effect_size_power import binary_effects, cohens_d, required_sample_size, two_sided_z_power
difference = 95 - 100
standardized = cohens_d(95, 100, 10)
print(f"Dispatch-time difference: {difference:.1f} minutes")
print(f"Standardized mean difference: {standardized:.3f}")
```

```python
risk_difference, risk_ratio = binary_effects(60, 1000, 80, 1000)
print(f"Late-shipment risk difference: {100 * risk_difference:.1f} percentage points")
print(f"Late-shipment risk ratio: {risk_ratio:.3f}")
```

```python
for size in [16, 32, 64, 128]:
    print(f"n={size:3d} per group: power={two_sided_z_power(0.5, size):.3f}")
```

```python
for effect in [0.2, 0.5, 0.8]:
    print(f"Effect {effect:.1f}: {required_sample_size(effect)} per group")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
sizes = list(range(5, 201))
for effect in [0.2, 0.5, 0.8]:
    axes[0].plot(sizes, [two_sided_z_power(effect, n) for n in sizes], label=f"d={effect}")
axes[0].axhline(0.8, color="black", linestyle="--")
axes[0].set(title="Prospective power curves", xlabel="Sample size per group", ylabel="Power")
axes[0].legend()
effects = [value / 100 for value in range(10, 101, 5)]
axes[1].plot(effects, [required_sample_size(effect) for effect in effects], color="#25705A")
axes[1].set(title="Sample size for 80% power", xlabel="Standardized effect", ylabel="Required per group")
fig.tight_layout()
plt.show()
```

```python
shipments, minutes_saved, value_per_minute = 200_000, 5, 2_000
gross = shipments * minutes_saved * value_per_minute
net = gross - 1_400_000_000
print(f"Expected gross value: Rp {gross / 1e9:.2f} billion")
print(f"Expected net value: Rp {net / 1e9:.2f} billion")
```
