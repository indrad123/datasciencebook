# Chapter 18: Probability Distributions
Synthetic NRG examples connect event processes with candidate probability models.

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from datasciencebook.probability_distributions import binomial_pmf, exponential_cdf, normal_cdf, poisson_pmf
print(f"Exactly 2 late among 5: {binomial_pmf(2, 5, .1):.4f}")
```

```python
print(f"Zero complaints at rate 2: {poisson_pmf(0, 2):.4f}")
```

```python
print(f"P(fill weight <= 115 g): {normal_cdf(115, 100, 15):.4f}")
print(f"P(wait <= 2 hours): {exponential_cdf(2, .5):.4f}")
```

```python
rng = np.random.default_rng(20260829)
late = rng.binomial(1, .1, 200)
print(f"Simulated late shipments: {late.sum()} of {len(late)}")
```

```python
k=np.arange(0,9); fig,axes=plt.subplots(1,2,figsize=(10,4))
axes[0].bar(k,[binomial_pmf(int(x),8,.2) for x in k],color="#287271"); axes[0].set(title="Binomial: late shipments",xlabel="Count",ylabel="Probability")
x=np.linspace(0,10,200); axes[1].plot(x,[exponential_cdf(v,.5) for v in x],color="#d1495b"); axes[1].set(title="Exponential waiting-time CDF",xlabel="Hours",ylabel="Cumulative probability")
fig.tight_layout(); plt.show()
```

```python
print(f"Normal model assigns some probability below zero: {normal_cdf(0, 10, 5):.4f}")
```
