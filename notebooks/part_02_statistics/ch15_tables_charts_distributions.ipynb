All NRG order and sales data are synthetic and generated with seed 20260829.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(20260829)
print("Synthetic NRG distribution study ready.")
```

```python
products = np.array(["Coffee", "Noodles", "Spices", "Tea"])
product = rng.choice(products, 300, p=[0.25, 0.35, 0.15, 0.25])
counts = {name: int(np.sum(product == name)) for name in products}
for name in products:
    print(f"{name}: count={counts[name]}, relative frequency={counts[name]/300:.3f}")
```

```python
late_probability = np.select(
    [product == "Coffee", product == "Noodles", product == "Spices", product == "Tea"],
    [0.16, 0.10, 0.20, 0.14],
)
late = rng.random(300) < late_probability
print(f"Late deliveries: {int(late.sum())} of 300")
for name in products:
    late_count = int(np.sum((product == name) & late))
    print(f"{name}: on time={counts[name]-late_count}, late={late_count}")
```

```python
delivery_days = np.clip(rng.lognormal(mean=0.9, sigma=0.45, size=300), 0.5, 10)
print(f"Mean delivery time: {delivery_days.mean():.2f} days")
print(f"Median delivery time: {np.median(delivery_days):.2f} days")
print(f"Delivered within 4 days: {np.mean(delivery_days <= 4):.1%}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
order = sorted(products, key=lambda name: counts[name])
axes[0].barh(order, [counts[name] for name in order], color="#2878B5")
axes[0].set(xlabel="Orders", title="Product frequency")
axes[1].hist(delivery_days, bins=np.arange(0, 10.5, 0.5), color="#F28E2B", edgecolor="white")
axes[1].set(xlabel="Delivery days", ylabel="Orders", title="Delivery-time distribution")
sorted_days = np.sort(delivery_days)
axes[2].step(sorted_days, np.arange(1, len(sorted_days)+1)/len(sorted_days), where="post")
axes[2].axvline(4, color="black", linestyle=":")
axes[2].set(xlabel="Delivery days", ylabel="Cumulative proportion", title="Empirical cumulative distribution")
fig.tight_layout()
print("Figure shows category frequencies, histogram bins, and the ECDF.")
```

```python
day = np.arange(1, 31)
daily_sales = np.rint(1200 + 18*np.arange(30) + 120*np.sin(np.arange(30)*2*np.pi/7) + rng.normal(0, 70, 30)).astype(int)
print(f"First five daily sales totals: {daily_sales[:5].tolist()}")
print(f"Average daily sales: {daily_sales.mean():.1f} units")
```

```python
fig2, ax = plt.subplots(figsize=(9, 4))
ax.plot(day, daily_sales, marker="o", markersize=3)
ax.set(xlabel="Day in synthetic month", ylabel="Sales units", title="NRG daily sales")
fig2.tight_layout()
print("Figure shows an ordered daily sales sequence.")
```
