All NRG examples are synthetic and generated with seed 20260829.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(20260829)
print("Synthetic NRG data-quality study ready.")
```

```python
print("Nominal: region and shipment mode")
print("Ordinal: satisfaction level")
print("Interval: Celsius temperature")
print("Ratio: delivery duration and weight")
```

```python
regions = np.array(["Java", "Sumatra", "Kalimantan", "Sulawesi"])
region = rng.choice(regions, 800, p=[0.55, 0.20, 0.13, 0.12])
base = np.select(
    [region == "Java", region == "Sumatra", region == "Kalimantan", region == "Sulawesi"],
    [2.0, 3.8, 5.2, 4.6],
)
duration = np.maximum(0.5, base + rng.normal(0, 0.7, 800))
counts = {name: int(np.sum(region == name)) for name in regions}
print("Synthetic population: 800 deliveries")
print(f"Population mean duration: {duration.mean():.3f} days")
print("Region counts: " + ", ".join(f"{name}={counts[name]}" for name in regions))
```

```python
java_indices = np.flatnonzero(region == "Java")
convenience_indices = rng.choice(java_indices, 100, replace=False)
stratum_indices = {
    name: rng.choice(np.flatnonzero(region == name), 25, replace=False) for name in regions
}
stratum_means = {name: duration[index].mean() for name, index in stratum_indices.items()}
weighted_mean = sum((counts[name] / 800) * stratum_means[name] for name in regions)
unweighted_mean = np.mean(list(stratum_means.values()))
print(f"Java-only convenience mean: {duration[convenience_indices].mean():.3f} days")
print(f"Unweighted equal-stratum mean: {unweighted_mean:.3f} days")
print(f"Population-weighted stratified mean: {weighted_mean:.3f} days")
```

```python
reference_weight = np.linspace(5, 50, 100)
biased_sensor = reference_weight + 0.4 + rng.normal(0, 0.25, 100)
unbiased_sensor = reference_weight + rng.normal(0, 0.25, 100)
biased_error = biased_sensor - reference_weight
unbiased_error = unbiased_sensor - reference_weight
print(f"Biased sensor average signed error: {biased_error.mean():.3f} kg")
print(f"Biased sensor mean absolute error: {np.mean(np.abs(biased_error)):.3f} kg")
print(f"Unbiased sensor average signed error: {unbiased_error.mean():.3f} kg")
print(f"Unbiased sensor mean absolute error: {np.mean(np.abs(unbiased_error)):.3f} kg")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].bar(regions, [counts[name] for name in regions], label="Population", alpha=0.65)
axes[0].bar(regions, [100, 0, 0, 0], label="Convenience sample", alpha=0.65)
axes[0].set(ylabel="Deliveries", title="Coverage of a convenience sample")
axes[0].tick_params(axis="x", rotation=20)
axes[0].legend()
axes[1].scatter(reference_weight, biased_error, s=14, alpha=0.65, label="Offset + noise")
axes[1].scatter(reference_weight, unbiased_error, s=14, alpha=0.65, label="Noise only")
axes[1].axhline(0, color="black", linestyle=":")
axes[1].set(xlabel="Reference weight (kg)", ylabel="Observed minus reference (kg)", title="Measurement error")
axes[1].legend()
fig.tight_layout()
print("Figure shows sampling imbalance and measurement-error patterns.")
```
