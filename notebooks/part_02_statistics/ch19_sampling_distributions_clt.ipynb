# Chapter 19: Sampling Distributions and the Central Limit Theorem
Repeated samples from a fixed synthetic NRG population reveal how sample means vary.

```python
import numpy as np
import matplotlib.pyplot as plt
from datasciencebook.sampling_distributions import finite_population_correction, standard_error
rng=np.random.default_rng(20260829)
population=rng.lognormal(4.5,.55,10000)
print(f"Population mean: {population.mean():.3f}")
print(f"Population SD: {population.std():.3f}")
```

```python
sampling_means={}
for n in [5,30,100]:
    means=rng.choice(population,size=(2000,n),replace=True).mean(axis=1)
    sampling_means[n]=means
    print(f"n={n}: mean={means.mean():.3f}, simulated SE={means.std(ddof=1):.3f}, theory={standard_error(population.std(),n):.3f}")
```

```python
fig,axes=plt.subplots(1,3,figsize=(12,3.5))
for ax,n in zip(axes,[5,30,100]):
    ax.hist(sampling_means[n],bins=25,color="#287271",edgecolor="white")
    ax.axvline(population.mean(),color="#d1495b")
    ax.set(title=f"Means for n={n}",xlabel="Sample mean",ylabel="Frequency")
fig.tight_layout();plt.show()
```

```python
print(f"SE at n=25: {standard_error(population.std(),25):.3f}")
print(f"SE at n=100: {standard_error(population.std(),100):.3f}")
```

```python
print(f"Finite population correction for N=400, n=100: {finite_population_correction(400,100):.4f}")
```
