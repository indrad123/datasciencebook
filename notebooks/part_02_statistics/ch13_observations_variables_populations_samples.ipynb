All NRG records below are synthetic and generated with seed 20260829.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(20260829)
print("Synthetic NRG population generator ready.")
```

```python
population_size = 400
store_id = np.array([f"ID-STORE-{i:03d}" for i in range(1, population_size + 1)])
regions = np.array(["Java", "Sumatra", "Kalimantan", "Sulawesi"])
formats = np.array(["Convenience", "Supermarket", "Kiosk"])
region = rng.choice(regions, population_size, p=[0.55, 0.20, 0.13, 0.12])
store_format = rng.choice(formats, population_size, p=[0.45, 0.40, 0.15])
print("Observational unit: one NRG retail store")
print(f"Population size: {population_size} stores")
print("Variables: store_id, region, format, weekly_sales_units")
```

```python
region_base = np.select(
    [region == "Java", region == "Sumatra", region == "Kalimantan", region == "Sulawesi"],
    [1250, 1050, 950, 900],
)
weekly_sales = np.rint(np.maximum(250, region_base + rng.normal(0, 180, population_size))).astype(int)
print(f"First five store IDs: {store_id[:5].tolist()}")
print(f"First five regions: {region[:5].tolist()}")
print(f"First five weekly sales: {weekly_sales[:5].tolist()}")
```

```python
sample_size = 40
sample_indices = rng.choice(population_size, sample_size, replace=False)
population_mean = weekly_sales.mean()
sample_mean = weekly_sales[sample_indices].mean()
print(f"Population parameter, mean weekly sales: {population_mean:.2f} units")
print(f"Sample size: {sample_size} stores")
print(f"Sampling fraction: {sample_size/population_size:.1%}")
print(f"Sample statistic, mean weekly sales: {sample_mean:.2f} units")
print(f"Estimation error for this sample: {sample_mean-population_mean:.2f} units")
```

```python
sample_means = np.array([
    weekly_sales[rng.choice(population_size, sample_size, replace=False)].mean()
    for _ in range(1000)
])
print("Repeated samples: 1,000")
print(f"Mean of sample means: {sample_means.mean():.2f} units")
print(f"Standard deviation of sample means: {sample_means.std(ddof=1):.2f} units")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(weekly_sales, bins=18, color="#2878B5", edgecolor="white")
axes[0].axvline(population_mean, color="black", linestyle=":", label="Population mean")
axes[0].set(xlabel="Weekly sales units", ylabel="Stores", title="Synthetic store population")
axes[0].legend()
axes[1].hist(sample_means, bins=20, color="#F28E2B", edgecolor="white")
axes[1].axvline(population_mean, color="black", linestyle=":", label="Population mean")
axes[1].set(xlabel="Sample mean", ylabel="Repeated samples", title="Sampling variation, n=40")
axes[1].legend()
fig.tight_layout()
print("Figure compares the population with repeated sample statistics.")
```
