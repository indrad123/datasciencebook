# Chapter 16: Centres, Quantiles, and Spread

This notebook uses a fixed seed and synthetic Nusantara Rasa Global delivery data. It compares measures of centre, position, and spread.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(20260829)
delivery_days = np.clip(rng.lognormal(mean=0.9, sigma=0.45, size=300), 0.5, 12)
print(f"Observations: {len(delivery_days)}")
print(f"Mean: {delivery_days.mean():.3f} days")
print(f"Median: {np.median(delivery_days):.3f} days")
```

```python
with_extreme_delay = np.append(delivery_days, 30.0)
print(f"Original mean: {delivery_days.mean():.3f}; with delay: {with_extreme_delay.mean():.3f}")
print(f"Original median: {np.median(delivery_days):.3f}; with delay: {np.median(with_extreme_delay):.3f}")
```

```python
five_numbers = np.quantile(delivery_days, [0, 0.25, 0.5, 0.75, 1])
print("Five-number summary:", [float(round(value, 3)) for value in five_numbers])
```

```python
sample_variance = np.var(delivery_days, ddof=1)
sample_sd = np.std(delivery_days, ddof=1)
iqr = np.quantile(delivery_days, 0.75) - np.quantile(delivery_days, 0.25)
print(f"Sample variance: {sample_variance:.3f} days squared")
print(f"Sample standard deviation: {sample_sd:.3f} days")
print(f"IQR: {iqr:.3f} days")
```

```python
regional_means = np.array([120.0, 150.0, 135.0])
store_counts = np.array([40, 10, 25])
print(f"Store-weighted mean: {np.average(regional_means, weights=store_counts):.1f} units")
print(f"Unweighted mean of region means: {regional_means.mean():.1f} units")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].hist(delivery_days, bins=20, color="#287271", edgecolor="white")
axes[0].axvline(delivery_days.mean(), color="#d1495b", label="Mean")
axes[0].axvline(np.median(delivery_days), color="#edae49", label="Median")
axes[0].set(title="Synthetic delivery times", xlabel="Days", ylabel="Orders")
axes[0].legend()
axes[1].boxplot([delivery_days, with_extreme_delay], tick_labels=["Original", "With 30-day delay"])
axes[1].set(title="Outlier changes the visible range", ylabel="Days")
fig.tight_layout()
plt.show()
```

```python
short_sample = np.array([10, 20, 30, 40])
print(f"Linear interpolation at p=0.25: {np.quantile(short_sample, 0.25):.1f}")
```
