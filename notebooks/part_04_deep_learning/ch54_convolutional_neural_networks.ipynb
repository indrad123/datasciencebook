Apply transparent filters and pooling to a synthetic NRG package image.


```python
from pathlib import Path
import sys,numpy as np,matplotlib.pyplot as plt
sys.path.insert(0,str(Path.cwd().parents[1]/'src'))
from datasciencebook.convolution import *
print('Imports ready.')
```

```python
image=np.zeros((12,12));image[3:9,3:9]=1;image[5:7,6:10]=.2;edge=np.array([[-1,0,1],[-1,0,1],[-1,0,1]]);feature=convolve2d(image,edge,padding=1)
print(f'Image={image.shape}; feature map={feature.shape}; strongest response={feature.max():.1f}')
```

```python
pooled=max_pool2d(np.abs(feature),2,2);print(f'Pooled shape={pooled.shape}; parameter example={conv_parameter_count(3,3,3,16)}')
```

```python
for p,s in [(0,1),(1,1),(1,2)]:print(f'padding={p}, stride={s}, output={output_size(32,3,s,p)}')
print(f'Receptive field={receptive_field([(3,1),(2,2),(3,1)])}')
```

```python
fig,axes=plt.subplots(1,2,figsize=(9,4));axes[0].imshow(image,cmap='gray');axes[0].set(title='Synthetic package patch');axes[1].imshow(feature,cmap='coolwarm');axes[1].set(title='Vertical-edge response');[a.axis('off') for a in axes];fig.tight_layout();plt.show()
```

## Interpretation

One shared filter detects vertical changes wherever they occur. Padding preserves size, pooling compresses the response, and stacked kernels enlarge the receptive field.
