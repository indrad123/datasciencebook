# Chapter 49: When Deep Learning Earns Its Cost

An NRG package-inspection proposal is assessed through incremental value, lifecycle cost, and readiness gates.


```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
sys.path.insert(0,str(Path.cwd().parents[1]/'src'))
from datasciencebook.deep_learning_value import annual_value,annual_cost,net_value,readiness_score,recommend
print('Imports ready.')
```

```python
cases=500_000;baseline_error=.08;deep_error=.06;value_per_correct=18_000
value=annual_value(cases,baseline_error,deep_error,value_per_correct)
cost=annual_cost(12,2_500_000,cases,35,190_000_000)
print(f'Gross incremental value: IDR {value:,.0f}; lifecycle cost: {cost:,.0f}; net: {net_value(value,cost):,.0f}')
```

```python
dims={'data':.82,'baseline':.90,'operations':.64,'governance':.76};ready=readiness_score(*dims.values());decision=recommend(net_value(value,cost),ready);print(f'Readiness={ready:.2f}; recommendation={decision}')
```

```python
gains=np.linspace(.005,.04,8);nets=[net_value(annual_value(cases,baseline_error,baseline_error-g,value_per_correct),cost) for g in gains];break_even=cost/(cases*value_per_correct);print(f'Break-even absolute error reduction: {break_even:.3f}')
```

```python
fig,axes=plt.subplots(1,2,figsize=(11,4));axes[0].bar(dims.keys(),dims.values());axes[0].axhline(.6,color='red',ls=':');axes[0].set(ylim=(0,1),title='Minimum readiness gate');axes[1].plot(gains,nets,'o-');axes[1].axhline(0,color='black',ls=':');axes[1].set(xlabel='Absolute error reduction',ylabel='Annual net value (IDR)',title='Value sensitivity');fig.tight_layout();plt.show()
```

## Interpretation

The proposed two-point error reduction does not repay the stated lifecycle cost, even though readiness clears the minimum gate. NRG should improve the baseline, reduce cost, find reusable value, or require a larger validated gain before piloting.
