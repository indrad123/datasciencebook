# Chapter 57: Deep Learning for Tables, Text, Images, and Time Series

Compare NRG model candidates across modalities, quality, and cost.


```python
from pathlib import Path
import sys,numpy as np,matplotlib.pyplot as plt
sys.path.insert(0,str(Path.cwd().parents[1]/'src'))
from datasciencebook.modality import *
print('Imports ready.')
```

```python
modalities=['Table','Text','Image','Time series'];baselines=np.array([.84,.86,.78,.81]);deep=np.array([.83,.90,.87,.84]);gains=[relative_improvement(a,b) for a,b in zip(baselines,deep)]
for m,g in zip(modalities,gains):print(f'{m}: relative gain={g:.1%}')
```

```python
hours=[2,14,28,18];latency=[3,18,35,12];scores=[value_per_compute(max(g,0),h,l) for g,h,l in zip(gains,hours,latency)]
print('Highest value-per-compute:',modalities[int(np.argmax(scores))])
print('Window count:',window_count(240,24,12))
```

```python
package=['P1','P1','P2','P3','P3'];train,test=leakage_safe_split(package,['P3'])
print('Train rows:',train.astype(int).tolist())
print('Test rows:',test.astype(int).tolist())
```

```python
fig,axes=plt.subplots(1,2,figsize=(10,4));x=np.arange(4);axes[0].bar(x-.18,baselines,.36,label='baseline');axes[0].bar(x+.18,deep,.36,label='deep');axes[0].set(xticks=x,xticklabels=modalities,ylim=(.7,.95),title='Held-out quality');axes[0].tick_params(axis='x',rotation=25);axes[0].legend();axes[1].scatter(hours,100*np.array(gains),s=np.array(latency)*5);[axes[1].annotate(m,(h,100*g)) for m,h,g in zip(modalities,hours,gains)];axes[1].axhline(0,color='grey',lw=1);axes[1].set(title='Gain and training cost',xlabel='Training hours',ylabel='Relative gain (%)');fig.tight_layout();plt.show()
```

## Interpretation

Deep learning is not uniformly superior. The tabular baseline wins in this synthetic comparison, while text gives the best gain per unit of compute and images give the largest raw gain.
