# Chapter 51: Layers, Activations, Losses, and Forward Propagation

Trace a small NRG network from sensor inputs to class probabilities.


```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
sys.path.insert(0,str(Path.cwd().parents[1]/'src'))
from datasciencebook.forward_network import activation,forward_network,multiclass_cross_entropy,parameter_count
print('Imports ready.')
```

```python
layers=[{'weights':np.array([[1,-1],[-1,1]]),'bias':np.zeros(2),'activation':'relu'},{'weights':np.array([[10],[10]]),'bias':np.array([-5]),'activation':'sigmoid'}]
X=np.array([[0,0],[0,1],[1,0],[1,1]],float); output,trace=forward_network(X,layers,True)
print(f'Shapes: input={X.shape}, hidden={trace[0]["activation"].shape}, output={output.shape}; parameters={parameter_count(layers)}')
```

```python
pred=(output[:,0]>=.5).astype(int); print(f'XOR probabilities={np.round(output[:,0],3).tolist()}; predictions={pred.tolist()}')
```

```python
rng=np.random.default_rng(51); logits=rng.normal(size=(8,3)); probs=activation(logits,'softmax'); labels=np.array([0,1,2,1,0,2,2,1]); loss=multiclass_cross_entropy(labels,probs)
print(f'Row sums valid={np.allclose(probs.sum(axis=1),1)}; multiclass loss={loss:.3f}')
```

```python
z=np.linspace(-5,5,300); fig,axes=plt.subplots(1,2,figsize=(11,4)); axes[0].plot(z,activation(z,'sigmoid'),label='sigmoid'); axes[0].plot(z,activation(z,'tanh'),label='tanh'); axes[0].plot(z,activation(z,'relu'),label='ReLU'); axes[0].legend(); axes[0].set(title='Activation functions'); axes[1].scatter(X[:,0],X[:,1],c=output[:,0],cmap='coolwarm',s=180); axes[1].set(xticks=[0,1],yticks=[0,1],title='Hidden features solve XOR',xlabel='Input 1',ylabel='Input 2'); fig.tight_layout(); plt.show()
```

## Interpretation

The two hidden ReLU units activate for opposite input differences. Their combination lets the output separate XOR, while stable softmax produces valid multiclass probabilities.
