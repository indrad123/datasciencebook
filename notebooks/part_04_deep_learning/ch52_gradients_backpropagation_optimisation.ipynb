# Chapter 52: Gradients, Backpropagation, and Optimisation

Verify a gradient, train XOR, and inspect optimization health.


```python
from pathlib import Path
import sys,numpy as np,matplotlib.pyplot as plt
sys.path.insert(0,str(Path.cwd().parents[1]/'src'))
from datasciencebook.backpropagation import *
print('Imports ready.')
```

```python
X=np.array([[0,0],[0,1],[1,0],[1,1]],float);y=np.array([0,1,1,0]);p=initialize(2,4,2);_,cache=forward(X,p);g=gradients(y,p,cache);num=finite_difference(X,y,p,'W1',(0,0));print(f'Analytical={g["W1"][0,0]:.6f}; numerical={num:.6f}; difference={abs(g["W1"][0,0]-num):.2e}')
```

```python
p,history,norms=train(X,y,4,.5,4000,2);prob=forward(X,p)[0];print(f'Initial loss={history[0]:.3f}; final loss={history[-1]:.4f}; predictions={(prob>=.5).astype(int).tolist()}')
```

```python
for rate in [.01,.1,.5]:
 _,h,_=train(X,y,4,rate,800,2);print(f'learning_rate={rate:.2f}, final_loss={h[-1]:.4f}')
```

```python
fig,axes=plt.subplots(1,2,figsize=(11,4));axes[0].plot(history);axes[0].set(xlabel='Epoch',ylabel='Loss',title='Optimization');axes[1].plot(norms);axes[1].set(xlabel='Epoch',ylabel='Gradient norm',title='Gradient health');fig.tight_layout();plt.show()
```

## Interpretation

The analytical and numerical derivatives agree. The network learns XOR, while the learning-rate comparison shows why optimization settings must be diagnosed rather than assumed.
