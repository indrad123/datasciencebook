# Chapter 50: From a Linear Model to an Artificial Neuron

Train one transparent neuron for NRG package-damage risk and expose its linear boundary.


```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
sys.path.insert(0,str(Path.cwd().parents[1]/'src'))
from datasciencebook.neuron import linear_output,sigmoid,train_logistic_neuron,neuron_probabilities,classify
print('Imports ready.')
```

```python
x=np.array([.8,.4]); w=np.array([1.8,1.1]); b=-1.4
z=linear_output(x,w,b); p=sigmoid(z)
print(f'Preactivation={z:.3f}; damage probability={p:.3f}')
```

```python
rng=np.random.default_rng(50); X=rng.normal(size=(240,2)); latent=1.6*X[:,0]+1.0*X[:,1]-.35+rng.normal(0,.55,240); y=(latent>0).astype(int)
fit=train_logistic_neuron(X,y,learning_rate=.15,epochs=900); pred=classify(neuron_probabilities(X,fit['weights'],fit['bias']))
print(f"Initial loss={fit['loss_history'][0]:.3f}; final loss={fit['loss_history'][-1]:.3f}; accuracy={np.mean(pred==y):.3f}")
```

```python
Xor=np.array([[0,0],[0,1],[1,0],[1,1]],float); yor=np.array([0,1,1,0]); xor_fit=train_logistic_neuron(Xor,yor,.2,1000); xor_pred=classify(neuron_probabilities(Xor,xor_fit['weights'],xor_fit['bias']))
print(f'XOR predictions={xor_pred.tolist()}; accuracy={np.mean(xor_pred==yor):.2f}')
```

```python
fig,axes=plt.subplots(1,2,figsize=(11,4)); axes[0].plot(fit['loss_history']); axes[0].set(xlabel='Epoch',ylabel='Cross-entropy',title='Training loss'); xx=np.linspace(X[:,0].min(),X[:,0].max()); yy=-(fit['weights'][0]*xx+fit['bias'])/fit['weights'][1]; axes[1].scatter(X[:,0],X[:,1],c=y,cmap='coolwarm',alpha=.7); axes[1].plot(xx,yy,color='black'); axes[1].set(xlabel='Impact score',ylabel='Moisture score',title="One neuron's linear boundary"); fig.tight_layout(); plt.show()
```

## Interpretation

Gradient descent lowers loss and learns a useful straight boundary for the simulated NRG data. More epochs cannot solve XOR with one linear boundary; that limitation motivates layers.
