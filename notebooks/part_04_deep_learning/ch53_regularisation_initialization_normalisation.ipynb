# Chapter 53: Regularisation, Initialization, and Normalisation

Compare signal scale, stored normalization, weight penalties, and dropout.


```python
from pathlib import Path
import sys,numpy as np,matplotlib.pyplot as plt
sys.path.insert(0,str(Path.cwd().parents[1]/'src'))
from datasciencebook.training_controls import *
print('Imports ready.')
```

```python
rng=np.random.default_rng(53);X=rng.normal(size=(1000,32));small=activation_variances(X,[64]*8,'small',1);xavier=activation_variances(X,[64]*8,'xavier',1)
print(f'Layer-8 variance: small={small[-1]:.8f}; xavier={xavier[-1]:.4f}')
```

```python
sensor=np.array([[2,20],[4,24],[6,28]],float);state=fit_standardizer(sensor);z=apply_standardizer(sensor,state);print(f'Standardized means={np.round(z.mean(0),6).tolist()}; scales={np.round(z.std(0),3).tolist()}')
```

```python
x=np.ones(20000);d=dropout(x,.25,True,7);penalty=l2_penalty([np.array([[1.,-2.],[.5,0]])],.1);print(f'Dropout mean={d.mean():.3f}; zero fraction={np.mean(d==0):.3f}; L2 penalty={penalty:.3f}')
```

```python
fig,axes=plt.subplots(1,2,figsize=(11,4));axes[0].plot(range(1,9),small,'o-',label='small');axes[0].plot(range(1,9),xavier,'o-',label='Xavier');axes[0].set(xlabel='Layer',ylabel='Activation variance',title='Signal propagation');axes[0].legend();B,_=batch_normalize(rng.normal(5,3,(300,4)));axes[1].hist(B[:,0],bins=25);axes[1].set(title='Batch-normalized activation',xlabel='Value');fig.tight_layout();plt.show()
```

## Interpretation

Tiny weights erase signal across layers, whereas Xavier initialization preserves useful variance. Stored input scaling, inverted dropout, and L2 penalties each solve a different problem.
