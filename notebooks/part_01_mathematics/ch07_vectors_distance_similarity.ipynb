```python
import math
import numpy as np
import matplotlib.pyplot as plt

def paired(a,b):
    if len(a)==0 or len(a)!=len(b):
        raise ValueError("vectors must have the same non-zero dimension")
    return np.asarray(a,dtype=float),np.asarray(b,dtype=float)

def euclidean_distance(a,b):
    left,right=paired(a,b)
    return float(np.sqrt(np.sum((left-right)**2)))

def cosine_similarity(a,b):
    left,right=paired(a,b)
    denominator=np.linalg.norm(left)*np.linalg.norm(right)
    if denominator==0:
        raise ValueError("cosine similarity requires non-zero vectors")
    return float(np.dot(left,right)/denominator)
```

## Situation 1: Warehouse demand profiles

```python
a=np.array([4,3,5])
b=np.array([2,4,6])
print(f"Combined profile: {(a+b).tolist()}")
print(f"A minus B: {(a-b).tolist()}")
print(f"Dot product: {int(np.dot(a,b))}")
```

## Situation 2: Vectors as arrows

```python
fig,ax=plt.subplots(figsize=(6,5))
origin=np.zeros(2)
for vector,label,color in [(np.array([3,2]),"u=(3,2)","tab:blue"),(np.array([1,4]),"v=(1,4)","tab:orange")]:
    ax.quiver(*origin,*vector,angles="xy",scale_units="xy",scale=1,color=color,label=label)
ax.set(xlim=(0,5),ylim=(0,5),xlabel="First component",ylabel="Second component",title="Two vector representations")
ax.set_aspect("equal")
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
```

The arrow direction and length present geometric information that complements the component representation.

## Situation 3: Distance and directional similarity

```python
print(f"Euclidean distance: {euclidean_distance(a,b):.3f}")
print(f"Manhattan distance: {np.abs(a-b).sum():.0f}")
print(f"Cosine similarity: {cosine_similarity(a,b):.3f}")
```

Distance describes separation. Cosine similarity describes directional alignment and discounts positive overall magnitude.

## Situation 4: Scale changes distance

```python
customer_a=np.array([10,2_000_000],dtype=float)
customer_b=np.array([15,4_000_000],dtype=float)
raw=np.linalg.norm(customer_a-customer_b)
scaled_a=np.array([10,2.0])
scaled_b=np.array([15,4.0])
print(f"Raw distance: {raw:.3f}")
print(f"Rescaled distance: {np.linalg.norm(scaled_a-scaled_b):.3f}")
```

The second calculation expresses spending in millions of rupiah. The comparison shows that distance depends on representation and units.

## Situation 5: Neighbour geometry

```python
grid=np.linspace(-3,3,241)
x,y=np.meshgrid(grid,grid)
euclidean=np.sqrt(x*x+y*y)
manhattan=np.abs(x)+np.abs(y)
fig,ax=plt.subplots(figsize=(6,5))
ax.contour(x,y,euclidean,levels=[2],colors=["tab:blue"])
ax.contour(x,y,manhattan,levels=[2],colors=["tab:orange"])
ax.plot([],[],color="tab:blue",label="Euclidean distance 2")
ax.plot([],[],color="tab:orange",label="Manhattan distance 2")
ax.set(xlabel="Component 1 difference",ylabel="Component 2 difference",title="Equal-distance boundaries")
ax.set_aspect("equal")
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
```

Different measures create different definitions of an equally distant neighbour.
