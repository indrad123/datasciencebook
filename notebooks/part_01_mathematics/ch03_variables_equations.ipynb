```python
from math import isclose

def evaluate_linear(a, x, b=0.0):
    return a * x + b

def solve_linear(a, b, c):
    if a == 0:
        raise ValueError("a must not be zero for a unique solution")
    return (c - b) / a

def solve_both_sides(a, b, c, d):
    coefficient = a - c
    constant = d - b
    if coefficient == 0:
        return ("infinite", None) if constant == 0 else ("none", None)
    return "unique", constant / coefficient
```

## Situation 1: Solve the pallet plan

```python
x = solve_linear(18, 120, 1200)
verified = isclose(evaluate_linear(18, x, 120), 1200, abs_tol=1e-9)
print(f"Cartons per pallet: {x:.0f}")
print(f"Equation verified: {verified}")
```

The value 60 is meaningful only after x is defined as cartons per pallet. Substitution verifies the arithmetic.

## Situation 2: Equal freight costs

```python
classification, quantity = solve_both_sides(4000, 300000, 6000, 180000)
cost = 300000 + 4000 * quantity
print(f"Classification: {classification}")
print(f"Equal-cost quantity: {quantity:.0f} units")
print(f"Cost of each plan: IDR {cost:,.0f}")
```

## Situation 3: Classify special cases

```python
print("Same expressions:", solve_both_sides(3, 6, 3, 6))
print("Contradictory expressions:", solve_both_sides(2, 10, 2, 13))
```

A non-unique result is information about the equation. It is not a software failure.

## Situation 4: Apply an operational constraint

```python
from math import ceil
algebraic_pallets = 42.3
whole_pallets = ceil(algebraic_pallets)
print(f"Algebraic result: {algebraic_pallets} pallets")
print(f"Minimum whole pallets: {whole_pallets}")
```

Rounding up is justified here only under the stated rule that the requirement must be met using whole pallets. Different operating rules could lead to a different action.
