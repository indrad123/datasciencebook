```python
import numpy as np
import matplotlib.pyplot as plt
```

## A constant shipment rate

```python
rate = 40  # cases per hour
hours = 6
print(f"Accumulated shipments: {rate * hours} cases")
```

## Left rectangles and trapezoids

```python
x = np.array([0., 1., 2., 3.])
y = 2*x
widths = np.diff(x)
left = np.sum(y[:-1] * widths)
trapezoids = np.sum((y[:-1] + y[1:]) * 0.5 * widths)
print(f"Left rectangles: {left:.1f}")
print(f"Trapezoids: {trapezoids:.1f}")
print(f"Exact integral: {3**2:.1f}")
```

## Refinement for a curved function

```python
for n in [2, 4, 20]:
    grid = np.linspace(0, 2, n + 1)
    values = grid**2
    estimate = np.sum((values[:-1] + values[1:]) * 0.5 * np.diff(grid))
    print(f"n={n:2d}: trapezoidal estimate={estimate:.6f}")
print(f"exact: {8/3:.6f}")
```

```python
fine = np.linspace(0, 2, 300)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(fine, fine**2, color="navy")
grid = np.linspace(0, 2, 5)
axes[0].bar(grid[:-1], grid[:-1]**2, width=np.diff(grid), align="edge", alpha=0.4, edgecolor="black")
axes[0].set_title("Left rectangles for x²")
time = np.array([0., 1., 2.5, 4., 6.])
flow = np.array([12., 9., 4.5, 0., -6.])
increments = (flow[:-1] + flow[1:]) * 0.5 * np.diff(time)
inventory = 100 + np.concatenate(([0.], np.cumsum(increments)))
axes[1].plot(time, inventory, marker="o")
axes[1].set_title("Cumulative inventory estimate")
axes[1].set_xlabel("hours"); axes[1].set_ylabel("cases")
axes[0].set_xlabel("x"); axes[0].set_ylabel("f(x)")
for ax in axes: ax.grid(alpha=0.25)
fig.tight_layout()
plt.show()
```

## Unequal observation intervals

```python
time = np.array([0., 1., 2.5, 4.])
flow = np.array([10., 12., 9., 6.])
widths = np.diff(time)
total = np.sum((flow[:-1] + flow[1:]) * 0.5 * widths)
print("Interval widths:", widths)
print(f"Trapezoidal accumulation: {total:.2f} cases")
```

## Net change versus total movement

```python
grid = np.linspace(0, 6, 6001)
net_flow = 12 - 3*grid
net_change = np.sum((net_flow[:-1] + net_flow[1:]) * 0.5 * np.diff(grid))
absolute_flow = np.abs(net_flow)
total_movement = np.sum((absolute_flow[:-1] + absolute_flow[1:]) * 0.5 * np.diff(grid))
print(f"Net inventory change: {net_change:.1f} cases")
print(f"Total absolute flow: {total_movement:.1f} cases")
```

## Interpretation and limitations

Numerical integration assumes behaviour between observations. Results can be misleading when timestamps are wrong, observations are missing, or sharp changes occur between samples. Preserve units and distinguish net accumulation from total absolute movement.
