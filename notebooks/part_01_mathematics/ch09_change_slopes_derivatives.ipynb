## Average order-arrival rate

```python
orders_start, orders_end = 240, 300
hour_start, hour_end = 9, 12
total_change = orders_end - orders_start
average_rate = total_change / (hour_end - hour_start)
print(f"Total change: {total_change} orders")
print(f"Average rate: {average_rate} orders per hour")
```

## Secant slopes approach a derivative

```python
square = lambda x: x**2
x0 = 3.0
for h in [1.0, 0.1, 0.01, 0.001]:
    slope = (square(x0 + h) - square(x0)) / h
    print(f"h={h:.3f}: slope={slope:.6f}")
```

```python
x = np.linspace(0.5, 5.0, 300)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(x, square(x), label="f(x)=x²")
for h in [1.5, 0.7, 0.2]:
    m = (square(x0 + h) - square(x0)) / h
    axes[0].plot(x, square(x0) + m*(x-x0), label=f"secant h={h}")
axes[0].scatter([x0], [square(x0)], color="black")
axes[0].set_title("Shrinking secant intervals")
axes[0].legend(fontsize=8)
axes[1].plot(x, square(x), label="f(x)=x²")
axes[1].plot(x, square(x0) + 6*(x-x0), label="tangent slope 6")
axes[1].scatter([x0], [square(x0)], color="black")
axes[1].set_title("Tangent at x=3")
axes[1].legend()
for ax in axes:
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.grid(alpha=0.25)
fig.tight_layout()
plt.show()
```

## Forward and central finite differences

```python
def forward_difference(function, point, step=1e-5):
    return (function(point + step) - function(point)) / step

def central_difference(function, point, step=1e-5):
    return (function(point + step) - function(point - step)) / (2*step)

print(f"Forward: {forward_difference(square, 3):.8f}")
print(f"Central: {central_difference(square, 3):.8f}")
print(f"Exact: {2*3:.8f}")
```

## Marginal production cost

```python
cost = lambda q: 0.02*q**2 + 15*q + 600
marginal_cost = lambda q: 0.04*q + 15
print(f"Cost at 100 cases: {cost(100):.1f}")
print(f"Marginal cost near 100 cases: {marginal_cost(100):.1f} per case")
```

## Derivative of a forecast-loss function

```python
loss = lambda w: (w - 4)**2 + 2
loss_derivative = lambda w: 2*(w - 4)
for w in [2, 4, 6]:
    print(f"w={w}: loss={loss(w)}, derivative={loss_derivative(w)}")
```

## Interpretation and limitations

The derivative describes local change. A positive derivative means the function is locally increasing; a negative derivative means it is locally decreasing. Numerical results depend on smoothness, the finite-difference step, floating-point precision, and the resolution of the underlying measurements.
