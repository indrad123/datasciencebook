```python
import numpy as np

def weighted_sum(values, weights):
    if len(values) == 0 or len(values) != len(weights):
        raise ValueError("values and weights must have the same non-zero length")
    return sum(value * weight for value, weight in zip(values, weights))

def sum_squared_errors(actual, predicted):
    if len(actual) == 0 or len(actual) != len(predicted):
        raise ValueError("actual and predicted must have the same non-zero length")
    return sum((observed-estimate)**2 for observed,estimate in zip(actual,predicted))
```

## Situation 1: Expand a summation

```python
terms=[2*i+1 for i in range(1,5)]
print(f"Terms: {terms}")
print(f"Total: {sum(terms)}")
```

`range(1, 5)` represents the inclusive mathematical indices 1 through 4 because Python excludes its stop value.

## Situation 2: NRG revenue as a weighted sum

```python
prices=[20,30,25]
quantities=[10,5,8]
print(f"Revenue: {weighted_sum(prices,quantities)} thousand rupiah")
```

## Situation 3: Product and region totals

```python
shipments=np.array([[10,12,8],[7,9,11]])
print(f"Grand total: {shipments.sum()}")
print(f"Product totals: {shipments.sum(axis=1).tolist()}")
print(f"Region totals: {shipments.sum(axis=0).tolist()}")
```

## Situation 4: Errors and conditional sums

```python
actual=[10,12,9]
predicted=[11,10,9]
values=[80,120,150,90]
selected=[value for value in values if value>100]
print(f"SSE: {sum_squared_errors(actual,predicted)}")
print(f"Values above 100: {selected}")
print(f"Conditional sum: {sum(selected)}")
print(f"Conditional count: {len(selected)}")
```

## Situation 5: Input protection

```python
try:
    weighted_sum([1,2],[1])
except ValueError as error:
    print(error)
```
