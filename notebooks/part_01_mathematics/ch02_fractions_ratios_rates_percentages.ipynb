# Chapter 2: Fractions, Ratios, Rates, and Percentages

**Level:** Foundation  
**Prerequisite:** Chapter 1  
**Estimated study time:** 60 to 90 minutes

Objectives: calculate fractions, unit rates, percentages, percentage changes, percentage-point changes, and weighted percentages; then interpret each result in context.

**Data statement:** All Nusantara Rasa Global examples are fictional and synthetic.

```python
from dataclasses import dataclass

def unit_rate(amount, reference):
    if reference == 0:
        raise ValueError("reference must not be zero")
    return amount / reference

def percentage(part, whole):
    return unit_rate(part, whole) * 100

def percentage_change(old, new):
    if old == 0:
        raise ValueError("old value must not be zero")
    return (new - old) / old * 100

def weighted_percentage(parts, wholes):
    if len(parts) != len(wholes):
        raise ValueError("parts and wholes must have the same length")
    return percentage(sum(parts), sum(wholes))
```

## Situation 1: Compare suppliers with a unit rate

```python
supplier_a = unit_rate(495000, 18)
supplier_b = unit_rate(675000, 25)
print(f"Supplier A: IDR {supplier_a:,.0f}/kg")
print(f"Supplier B: IDR {supplier_b:,.0f}/kg")
```

Supplier B has the lower price per kilogram. This comparison concerns price only; it does not include quality or delivery reliability.

## Situation 2: Percentage and percentage change

```python
print(f"On-time deliveries: {percentage(1170, 1250):.1f}%")
print(f"Demand change: {percentage_change(800, 920):.1f}%")
```

The denominator for the delivery percentage is all deliveries. The denominator for percentage change is the original demand.

## Situation 3: Combine percentages with their counts

```python
accepted = [960, 1425]
inspected = [1000, 1500]
line_rates = [percentage(p, w) for p, w in zip(accepted, inspected)]
print(f"Simple average of line rates: {sum(line_rates)/len(line_rates):.1f}%")
print(f"Correct combined rate: {weighted_percentage(accepted, inspected):.1f}%")
```

The correct combined rate uses the underlying counts. A simple average gives each line equal influence even though the lines inspected different numbers of packs.

## Limitation check: an original value of zero

```python
try:
    percentage_change(0, 25)
except ValueError as error:
    print(error)
```

When the old value is zero, report the absolute change and context rather than an undefined percentage change.
