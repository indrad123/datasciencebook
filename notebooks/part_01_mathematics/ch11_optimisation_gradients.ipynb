```python
import numpy as np
import matplotlib.pyplot as plt
```

## Trial-and-error grid search

```python
loss = lambda w: (w - 4)**2 + 2
gradient = lambda w: 2*(w - 4)
candidates = np.array([0., 2., 4., 6.])
losses = loss(candidates)
best = np.argmin(losses)
print("Candidates:", candidates)
print("Losses:", losses)
print(f"Best candidate: {candidates[best]} with loss {losses[best]}")
```

## Gradient descent step by step

```python
w = 0.0
learning_rate = 0.1
print(f"start: w={w:.4f}, loss={loss(w):.4f}")
for step in range(1, 6):
    w -= learning_rate * gradient(w)
    print(f"step {step}: w={w:.4f}, loss={loss(w):.4f}")
```

## Learning-rate behaviour

```python
rates = [0.02, 0.1, 0.8]
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
grid = np.linspace(-1, 9, 300)
axes[0].plot(grid, loss(grid))
for rate in rates:
    w = 0.0; path = [w]
    for _ in range(12):
        w -= rate * gradient(w); path.append(w)
    axes[0].plot(path, loss(np.array(path)), marker="o", label=f"rate={rate}")
    axes[1].plot([loss(v) for v in path], marker="o", label=f"rate={rate}")
axes[0].set_title("Parameter paths"); axes[0].set_xlabel("w"); axes[0].set_ylabel("loss")
axes[1].set_title("Loss by iteration"); axes[1].set_xlabel("iteration"); axes[1].set_ylabel("loss")
for ax in axes: ax.grid(alpha=0.25); ax.legend()
fig.tight_layout(); plt.show()
```

## A two-parameter gradient

```python
point = np.array([0., 0.])
grad = np.array([2*(point[0]-2), 2*(point[1]+1)])
updated = point - 0.25*grad
print("Gradient at (0, 0):", grad)
print("After one step:", updated)
```

## Stop when the gradient is small

```python
w = 0.0
iterations = 0
while abs(gradient(w)) > 1e-7 and iterations < 1000:
    w -= 0.2*gradient(w)
    iterations += 1
print(f"Iterations: {iterations}")
print(f"Final w: {w:.8f}")
print(f"Final loss: {loss(w):.8f}")
print(f"Gradient magnitude: {abs(gradient(w)):.8f}")
```

```python
history = []
w = 0.0
for _ in range(20):
    history.append(loss(w)); w -= 0.1*gradient(w)
print("Training objective decreased:", all(b <= a for a, b in zip(history, history[1:])))
print("Validation still requires a separate check.")
```

## Interpretation and limitations

A gradient supplies local direction, not a guarantee of global optimality. Learning rate, scaling, curvature, noise, initialisation, constraints, and the quality of the objective all affect the result.
