## Reading a matrix

```python
weekly_orders = np.array([[18, 22, 15], [14, 17, 20]])
print("Shape:", weekly_orders.shape)
print("Transpose:")
print(weekly_orders.T)
```

## Solve an NRG production system

Two production lines use 100 machine-hours and 90 labour-hours. The coefficient matrix describes the hours required by one batch of each product.

```python
A = np.array([[2.0, 1.0], [1.0, 3.0]])
b = np.array([100.0, 90.0])
batches = np.linalg.solve(A, b)
residual = A @ batches - b
print("Batches:", batches)
print("Residual:", residual)
print("Determinant:", round(np.linalg.det(A), 3))
print("Condition number:", round(np.linalg.cond(A), 3))
```

## Matrix multiplication combines compatible transformations

```python
resource_per_route = np.array([[3, 2], [1, 4]])
routes_per_week = np.array([[12, 5], [7, 9]])
print(resource_per_route @ routes_per_week)
```

## Classify a system by rank

```python
systems = {
    "unique": (np.array([[1., 1.], [2., -1.]]), np.array([5., 1.])),
    "no solution": (np.array([[1., 1.], [2., 2.]]), np.array([5., 11.])),
    "infinitely many": (np.array([[1., 1.], [2., 2.]]), np.array([5., 10.])),
}
for label, (matrix, outcomes) in systems.items():
    rank_a = np.linalg.matrix_rank(matrix)
    rank_augmented = np.linalg.matrix_rank(np.column_stack([matrix, outcomes]))
    print(f"{label}: rank(A)={rank_a}, rank([A|b])={rank_augmented}")
```

## See the three geometric cases

```python
x = np.linspace(-1, 7, 200)
cases = [
    ("Unique", 5 - x, 2*x - 1),
    ("No solution", 5 - x, 5.5 - x),
    ("Infinitely many", 5 - x, 5 - x),
]
fig, axes = plt.subplots(1, 3, figsize=(12, 3.5), sharex=True, sharey=True)
for ax, (title, y1, y2) in zip(axes, cases):
    ax.plot(x, y1, label="equation 1")
    ax.plot(x, y2, linestyle="--", label="equation 2")
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.grid(alpha=0.25)
axes[0].set_ylabel("y")
axes[0].legend()
fig.tight_layout()
plt.show()
```

## Protect a direct solve from a singular matrix

```python
singular = np.array([[1., 1.], [2., 2.]])
try:
    np.linalg.solve(singular, np.array([5., 10.]))
except np.linalg.LinAlgError:
    print("Direct solve stopped: the coefficient matrix is singular.")
```
