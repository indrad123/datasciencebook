# Chapter 12: Probability, Conditional Probability, and Bayes' Theorem
This notebook uses a fixed seed and synthetic Nusantara Rasa Global examples.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(20260829)
print("Reproducible random generator ready.")
```

```python
sample_space = set(range(1, 7))
even = {2, 4, 6}
print(f"Sample space: {sorted(sample_space)}")
print(f"P(even) = {len(even) / len(sample_space):.3f}")
```

```python
rolls = rng.integers(1, 7, size=100_000)
six_count = int(np.sum(rolls == 6))
print(f"Sixes in 100,000 trials: {six_count}")
print(f"Simulated P(six) = {np.mean(rolls == 6):.4f}")
print(f"Theoretical P(six) = {1/6:.4f}")
```

```python
p_late, p_damaged, p_both = 0.10, 0.12, 0.04
print(f"P(damaged | late) = {p_both / p_late:.3f}")
print(f"P(late or damaged) = {p_late + p_damaged - p_both:.3f}")
```

```python
defective = 20
sound = 980
defective_alerts = int(defective * 0.90)
sound_alerts = int(sound * 0.05)
total_alerts = defective_alerts + sound_alerts
print(f"Defective cartons that alert: {defective_alerts}")
print(f"Sound cartons that alert: {sound_alerts}")
print(f"P(defect | alert) = {defective_alerts}/{total_alerts} = {defective_alerts/total_alerts:.4f}")
```

```python
trial_counts = np.arange(1, len(rolls) + 1)
running = np.cumsum(rolls == 6) / trial_counts
priors = np.linspace(0.001, 0.20, 300)
posteriors = 0.90 * priors / (0.90 * priors + 0.05 * (1 - priors))
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(trial_counts, running, linewidth=0.8)
axes[0].axhline(1/6, color='black', linestyle=':')
axes[0].set(xlabel='Trials', ylabel='Running proportion of sixes', title='Simulation convergence')
axes[1].plot(priors, posteriors)
axes[1].scatter([0.02], [18/67], color='crimson', zorder=3)
axes[1].set(xlabel='Prior defect probability', ylabel='Posterior after alert', title='Base rates change the posterior')
fig.tight_layout()
print("Figure shows simulation convergence and the effect of the prior.")
```

```python
expected_loss = 0 * 0.8 + 100 * 0.2
new_posterior = 0.95 * 0.02 / (0.95 * 0.02 + 0.10 * 0.98)
print(f"Expected shipment loss = IDR {expected_loss:.1f} million")
print(f"Higher-alert detector posterior = {new_posterior:.4f}")
```
