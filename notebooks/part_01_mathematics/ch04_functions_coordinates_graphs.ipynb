**Data statement:** All Nusantara Rasa Global examples are fictional and synthetic.

```python
import numpy as np
import matplotlib.pyplot as plt

def linear_value(x, slope, intercept=0.0):
    return slope * x + intercept

def slope_between(x1, y1, x2, y2):
    if x2 == x1:
        raise ValueError("x coordinates must be different")
    return (y2-y1)/(x2-x1)

def line_intersection(m1,b1,m2,b2):
    if m1 == m2:
        raise ValueError("no unique intersection")
    x=(b2-b1)/(m1-m2)
    return x, linear_value(x,m1,b1)
```

## Situation 1: One function in four representations

```python
quantities=np.array([0,50,100,150])
costs=linear_value(quantities,4000,300000)
points=list(zip(quantities.tolist(),costs.tolist()))
print(points)
```

```python
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(quantities,costs,marker="o",label="C(q) = 300,000 + 4,000q")
ax.set(xlabel="Shipment quantity (cases)",ylabel="Freight cost (IDR)",title="NRG freight cost by shipment quantity")
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
fig
```

The same relationship appears as a rule, a table of values, ordered pairs, and a graph. Axis labels preserve units.

## Situation 2: Interpret slope

```python
m=slope_between(50,500000,100,700000)
print(f"Slope: IDR {m:,.0f} per case")
```

## Situation 3: Compare two freight plans

```python
q=np.arange(0,121,10)
plan_a=linear_value(q,4000,300000)
plan_b=linear_value(q,6000,180000)
cross_q,cross_cost=line_intersection(4000,300000,6000,180000)
print(f"Intersection: {cross_q:.0f} cases, IDR {cross_cost:,.0f}")
fig,ax=plt.subplots(figsize=(7,4))
ax.plot(q,plan_a,label="Plan A: 300,000 + 4,000q")
ax.plot(q,plan_b,label="Plan B: 180,000 + 6,000q")
ax.scatter([cross_q],[cross_cost],color="black",zorder=3)
ax.set(xlabel="Shipment quantity (cases)",ylabel="Freight cost (IDR)",title="Break-even comparison of freight plans")
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
```

Plan B begins lower but rises faster. The intersection shows equal modelled cost, not equal service quality.
