```python
from datasciencebook.decision_reasoning import (
    choose_intervention, expected_loss_no_action, intervention_threshold,
    realized_cost, select_with_capacity,
)
intervention_cost, late_loss = 12, 80
threshold = intervention_threshold(intervention_cost, late_loss)
print(f'Decision threshold: {threshold:.2f}')
print(f'Expected loss without action at p=0.20: {expected_loss_no_action(0.20, late_loss):.2f}')
print(f'Expedite at p=0.20: {choose_intervention(0.20, intervention_cost, late_loss)}')
```

```python
probabilities = [0.10, 0.80, 0.30, 0.60]
outcomes = [0, 1, 1, 0]
actions = select_with_capacity(probabilities, capacity=2)
selected = [index for index, action in enumerate(actions) if action]
policy_cost = realized_cost(outcomes, actions, intervention_cost, late_loss)
no_action_cost = realized_cost(outcomes, [False] * len(outcomes), intervention_cost, late_loss)
print(f'Selected shipment indexes: {selected}')
print(f'Realized cost with capacity policy: {policy_cost:.2f}')
print(f'Realized cost with no intervention: {no_action_cost:.2f}')
```

```python
rare_outcomes = [1] * 5 + [0] * 95
all_on_time = [0] * 100
accuracy = sum(actual == predicted for actual, predicted in zip(rare_outcomes, all_on_time)) / 100
detected = sum(actual == predicted == 1 for actual, predicted in zip(rare_outcomes, all_on_time))
print(f'All-on-time accuracy: {accuracy:.2f}')
print(f'Late shipments detected: {detected} of {sum(rare_outcomes)}')
```

```python
import matplotlib.pyplot as plt
grid = [value / 100 for value in range(101)]
loss_no_action = [expected_loss_no_action(value, late_loss) for value in grid]
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(grid, loss_no_action, label='Do not expedite')
ax.axhline(intervention_cost, color='tab:orange', label='Expedite')
ax.axvline(threshold, color='tab:gray', linestyle='--', label=f'Threshold = {threshold:.2f}')
ax.set(xlabel='Predicted late probability', ylabel='Expected loss', title='NRG simplified action comparison')
ax.legend()
fig.tight_layout()
fig
```

## Interpretation and limitations
The threshold follows from simplified fixed costs and assumes expediting avoids the full late loss. A real policy must estimate intervention effectiveness, verify probability calibration, respect capacity and route constraints, and monitor consequences by customer and region.
