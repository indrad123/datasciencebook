# Chapter 32: Feature Engineering

This notebook creates time-safe features for a small NRG weekly demand series.


```python
import matplotlib.pyplot as plt
from datasciencebook.features import (
    safe_ratio, cyclic_encode, lag, trailing_mean, fit_categories,
    one_hot, interaction, feature_record,
)
```

## 1. Create ratios and interactions


```python
print('Fill rate:', safe_ratio(920, 1000))
print('Undefined ratio:', safe_ratio(10, 0))
print('Promotion interaction:', interaction(1, 0.15))
```

## 2. Encode a calendar cycle


```python
for month in [1, 6, 12]:
    sine, cosine = cyclic_encode(month - 1, 12)
    print(month, round(sine, 3), round(cosine, 3))
```

## 3. Construct lagged and trailing features

Every value uses history strictly before the current week.


```python
demand = [100, 120, 90, 140, 150, 130]
lag_1 = lag(demand, 1)
mean_3 = trailing_mean(demand, 3)
print('Demand:', demand)
print('Lag 1:', lag_1)
print('Trailing mean 3:', mean_3)
```

## 4. Fit and apply a category vocabulary


```python
categories = fit_categories(['Jakarta', 'Dubai', 'Jakarta'])
print('Vocabulary:', categories)
print('Known:', one_hot('Jakarta', categories))
print('Unseen:', one_hot('Surabaya', categories))
```

## 5. Visualise time-safe demand features


```python
weeks = list(range(1, len(demand) + 1))
plt.plot(weeks, demand, marker='o', label='Current demand')
plt.plot(weeks, lag_1, marker='s', label='Lag 1')
plt.plot(weeks, mean_3, marker='^', label='Prior 3-week mean')
plt.xlabel('Week')
plt.ylabel('Cartons')
plt.title('NRG demand and historical features')
plt.legend()
plt.tight_layout()
plt.show()
```

## 6. Register the feature definition


```python
record = feature_record(
    'demand_mean_3w',
    'mean of three completed prior product-warehouse weeks',
    'Monday forecast origin',
    unit='cartons per week',
)
print(record)
```
