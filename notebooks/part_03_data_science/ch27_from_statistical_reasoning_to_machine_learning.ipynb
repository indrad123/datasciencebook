Synthetic NRG examples; no external data or private service is used.

```python
from datasciencebook.ml_bridge import mean_baseline, predict_constant, mean_absolute_error
training_demand = [100, 110, 90, 120]
test_demand = [108, 130]
baseline = mean_baseline(training_demand)
baseline_predictions = predict_constant(baseline, len(test_demand))
print(f'Baseline: {baseline:.1f}')
print(f'Test MAE: {mean_absolute_error(test_demand, baseline_predictions):.1f}')
```

```python
from datasciencebook.ml_bridge import fit_simple_line, predict_line
x_train = [20, 24, 30, 36, 40]
y_train = [26, 29, 37, 45, 47]
x_test = [28, 44]
y_test = [36, 50]
intercept, slope = fit_simple_line(x_train, y_train)
line_predictions = predict_line(x_test, intercept, slope)
constant = mean_baseline(y_train)
print(f'Line: actual_hours = {intercept:.2f} + {slope:.2f} * promised_hours')
print(f'Baseline MAE: {mean_absolute_error(y_test, predict_constant(constant, 2)):.2f}')
print(f'Line MAE: {mean_absolute_error(y_test, line_predictions):.2f}')
```

```python
import matplotlib.pyplot as plt
grid = list(range(18, 47))
fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(x_train, y_train, label='Training shipments')
ax.scatter(x_test, y_test, marker='s', label='Unseen test shipments')
ax.plot(grid, predict_line(grid, intercept, slope), color='tab:orange', label='Learned line')
ax.axhline(constant, color='tab:gray', linestyle='--', label='Mean baseline')
ax.set(xlabel='Promised transit hours', ylabel='Actual transit hours', title='NRG: baseline and learned model')
ax.legend()
fig.tight_layout()
fig
```

## Interpretation and limitations
The learned line improves MAE for these two synthetic test cases. This small example demonstrates workflow, not general deployment validity. A real evaluation needs enough future observations, stable feature availability, subgroup error checks, and comparison under operational costs.
