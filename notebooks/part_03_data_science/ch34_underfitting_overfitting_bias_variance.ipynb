# Chapter 34: Underfitting, Overfitting, Bias, and Variance

This notebook diagnoses NRG demand models using training and validation behaviour.


```python
import matplotlib.pyplot as plt
from datasciencebook.fit_diagnostics import (
    mean_squared_error, generalization_gap, bias_variance_at_point,
    select_by_validation, early_stopping_epoch, diagnose_losses,
)
```

## 1. Compare candidate complexity settings


```python
candidates = [
    {'name': 'degree 1', 'training_loss': 520, 'validation_loss': 560},
    {'name': 'degree 3', 'training_loss': 210, 'validation_loss': 240},
    {'name': 'degree 6', 'training_loss': 90, 'validation_loss': 230},
    {'name': 'degree 12', 'training_loss': 5, 'validation_loss': 480},
]
for item in candidates:
    item['gap'] = generalization_gap(item['training_loss'], item['validation_loss'])
    print(item)
print('Selected:', select_by_validation(candidates)['name'])
```

## 2. Visualise training and validation curves


```python
degrees = [1, 3, 6, 12]
plt.plot(degrees, [c['training_loss'] for c in candidates], marker='o', label='Training MSE')
plt.plot(degrees, [c['validation_loss'] for c in candidates], marker='s', label='Validation MSE')
plt.xlabel('Polynomial degree')
plt.ylabel('Mean squared error')
plt.title('NRG validation curve')
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Decompose error at one input


```python
decomposition = bias_variance_at_point([8, 10, 12], truth=11, noise_variance=2)
print(decomposition)
```

## 4. Use simple loss diagnostics cautiously


```python
print('High losses:', diagnose_losses(0.60, 0.65, acceptable_loss=0.40, gap_tolerance=0.20))
print('Large gap:', diagnose_losses(0.10, 0.50, acceptable_loss=0.40, gap_tolerance=0.20))
print('No simple signal:', diagnose_losses(0.20, 0.25, acceptable_loss=0.40, gap_tolerance=0.20))
```

## 5. Inspect illustrative learning curves


```python
sizes = [100, 250, 500, 1000, 2000]
train_loss = [0.08, 0.10, 0.12, 0.14, 0.15]
validation_loss = [0.52, 0.40, 0.32, 0.26, 0.22]
plt.plot(sizes, train_loss, marker='o', label='Training loss')
plt.plot(sizes, validation_loss, marker='s', label='Validation loss')
plt.xlabel('Training observations')
plt.ylabel('Loss')
plt.title('Illustrative high-variance learning curve')
plt.legend()
plt.tight_layout()
plt.show()
```

## 6. Select an early-stopping checkpoint


```python
losses = [0.80, 0.60, 0.50, 0.52, 0.55, 0.57]
print(early_stopping_epoch(losses, patience=1))
print('Example MSE:', mean_squared_error([100, 120, 90], [105, 110, 95]))
```
