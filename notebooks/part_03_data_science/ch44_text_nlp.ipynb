# Chapter 44: Text Data and Introductory NLP

Synthetic NRG distributor messages demonstrate TF-IDF, linear classification, error analysis, and multilingual limits.


```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
sys.path.insert(0,str(Path.cwd().parents[1]/'src'))
from datasciencebook.text_nlp import fit_text_classifier,predict_text,classification_summary,top_coefficients,tokenize
print('Imports ready.')
```

```python
service=['delivery late warehouse delay','shipment damaged and late','order missing please investigate','barang rusak saat tiba','pengiriman terlambat lagi','wrong quantity received']
ordinary=['delivery received thank you','order arrived on time','please send product catalogue','terima kasih barang diterima','need price list coffee','confirm next regular order']
texts=[];labels=[]
for i in range(12):
 for t in service:texts.append(t+f' case{i}');labels.append(1)
 for t in ordinary:texts.append(t+f' note{i}');labels.append(0)
idx=np.arange(len(texts));rng=np.random.default_rng(44);rng.shuffle(idx);cut=108;tr=idx[:cut];te=idx[cut:]
print(f'Train messages: {len(tr)}; test messages: {len(te)}; positive test: {sum(labels[i] for i in te)}')
```

```python
v,m=fit_text_classifier([texts[i] for i in tr],[labels[i] for i in tr]);pred=predict_text(v,m,[texts[i] for i in te]);summary=classification_summary([labels[i] for i in te],pred)
print(f"Accuracy={summary['accuracy']:.3f}; macro_F1={summary['macro_f1']:.3f}; vocabulary={len(v.vocabulary_)}")
```

```python
positive,negative=top_coefficients(v,m,6);print('Service-signal terms:',positive);print('Ordinary-signal terms:',negative);print('Tokens:',tokenize('Pengiriman terlambat, order #42'))
```

```python
coef=m.coef_[0];names=np.asarray(v.get_feature_names_out());order=np.argsort(np.abs(coef))[-12:];fig,axes=plt.subplots(1,2,figsize=(11,4));axes[0].barh(names[order],coef[order]);axes[0].set(title='Largest linear coefficients');axes[1].hist([len(tokenize(t)) for t in texts],bins=8);axes[1].set(xlabel='Tokens',ylabel='Messages',title='Document-length audit');fig.tight_layout();plt.show()
```

## Interpretation

Perfect performance reflects deliberately simple synthetic templates, not production readiness. Real evaluation needs time-separated, deduplicated, multilingual messages, stable label guidance, subgroup results, and human review of costly errors.
